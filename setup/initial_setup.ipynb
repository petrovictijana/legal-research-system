```sql
-- 1. Create schema
CREATE SCHEMA IF NOT EXISTS workspace.legal_research_system;
```

```sql
-- 2. Documents table
CREATE TABLE IF NOT EXISTS workspace.legal_research_system.documents (
    id STRING NOT NULL,
    filename STRING NOT NULL,
    file_path STRING NOT NULL,
    title STRING,
    status STRING NOT NULL,
    created_at TIMESTAMP NOT NULL,
    updated_at TIMESTAMP NOT NULL
)
USING DELTA;
```

```sql
-- 3. Document chunks table
CREATE TABLE IF NOT EXISTS workspace.legal_research_system.document_chunks (
    id STRING NOT NULL,
    document_id STRING NOT NULL,
    chunk_type STRING,
    article_number STRING,
    chunk_index INT NOT NULL,
    content STRING NOT NULL,
    metadata STRING,
    embedding ARRAY<FLOAT>,
    created_at TIMESTAMP NOT NULL
)
USING DELTA
TBLPROPERTIES (
    'delta.enableChangeDataFeed' = 'true'
);
```

```sql
ALTER TABLE workspace.legal_research_system.documents 
ADD COLUMNS (
    structured_content_path STRING
)
```

```sql
-- 4. Create volumes
CREATE VOLUME IF NOT EXISTS workspace.legal_research_system.raw_documents;
CREATE VOLUME IF NOT EXISTS workspace.legal_research_system.processed_documents;
```
